#Tarea Introducción a la Ciencia de Datos

```python

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = pd.read_csv("titanic.csv")

df.head()

```

#Variables




1.   PassengerId: Es un dato numérico, cualitativo, pero de tipo nominal ya que es un identificador
2.   Survived: Es un dato numérico, cualitativo y dicotómica
3. Pclass: Es un dato numérico, cualitativo de tipo ordinal, se puede ordenar
4. Name: Es un dato object (texto), cualitativo de tipo nominal
5. Sex: Es un dato object (texto), cualitativo de tipo dicotómico ( sea hombre o mujer)
6. Age: Es un dato numérico decimal (float), cuantitativo de tipo continuo
7. SibSp: Es un dato numérico entero, cuantitativo y discreto
8. Parch: Es un dato numérico entero, cuantitativo y discreto
9. Ticket: Es un dato object(texto), de tipo cualitativo y nominal aunque mezcla
números y letras por lo que podriamos incluirlo en alfanumérico
10. Fare: Es un dato numérico decimal (float), cuantitativo y continuo
11. Cabin: Es un dato object (texto), cualitativo de tipo nominal
12. Embarked:  Es un dato object (texto), cualitativo de tipo nominal





#Descripción



1. PassengerId: Identificador con un valor único asignado a cada pasajero
2. Survived: Nos dice si el pasajero sobrevivió o no ( Sí = 1, No = 0)
3. Pclass: Clase del boleto socioeconómico
4. Name: Nombre completo y título del caballero ( Mr., Mrs., etc.)
5. Sex: Sexo del pasajero
6. Age: Edad del pasajero
7. SibSp: Número de hermanos o cónyuges a bordo
8. Parch: Numero de padres e hijos a bordo
9. Ticket: Código o número del boleto
10. Fare: Tarifa del pasaje
11. Cabin: Número de cabina o camarote asignado
12. Embarked: Puerto en el que abordó el pasajero

```python
categoricas_c = ["Survived", "Pclass", "Sex", "SibSp", "Parch","Embarked"]
for columna in categoricas_c:
  conteo = df[columna].value_counts(dropna=False)
  print(f"Conteo de la columna {columna}:", conteo)

  plt.figure (figsize=(6,4))
  conteo.plot(kind="bar")
  plt.title(f"Conteo de {columna}")
  plt.xlabel(columna)
  plt.ylabel("Frecuencia")
  plt.show()
```

```python
numericas_c = ["Age", "Fare"]
for columna in numericas_c:
  plt.figure(figsize=(6,4))
  df[columna].dropna().plot.hist(bins=20, color='#2ca02c', edgecolor='black', alpha=0.8)
  plt.title(f"Histograma de {columna}")
  plt.xlabel(columna)
  plt.ylabel("Frecuencia")
  plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.kdeplot(x = df [df["Survived"] == 1]["Age"], label = "Sobrevivió", color = "green")
sns.kdeplot(x = df [df["Survived"] == 0]["Age"], label = "No sobrevivió", color = "red")
plt.title("Distribución de Edad según la supervivencia")
plt.xlabel("Edad")
plt.ylabel("Densidad")
plt.legend()
plt.grid(True, linestyle = "--", alpha = 0.7)
plt.show()
```
